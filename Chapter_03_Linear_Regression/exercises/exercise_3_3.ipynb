```python
import math
from matplotlib import pyplot as plt
import numpy as np
import random

# s
features = [100, 200, 200, 250, 325]
# p
labels = [200, 475, 400, 520,735]

def prediction_func(s):
    return 2 * s + 50
```

```python
# predictions

predictions = [prediction_func(s) for s in features]
predictions
```

```python
# MAE (Mean Absolute Error)

sum([abs(label - predictions[i]) for i, label in enumerate(labels)]) / len(labels)
```

```python
# MSE (Mean Squared Error)

math.sqrt(sum([(label - predictions[i]) ** 2 for i, label in enumerate(labels)]) / len(labels))
```
